# download data




```python
import pandas as pd
# website = https://celestrak.org/NORAD/elements/
oneweb_url = (
    "https://celestrak.org/NORAD/elements/"
    "gp.php?GROUP=ONEWEB&FORMAT=CSV"
)

oneweb_df = pd.read_csv(oneweb_url)

oneweb_df.head()
```

```python
print("Number of satellites:", oneweb_df.shape[0])
print("Number of columns:", oneweb_df.shape[1])
```

```python
print(oneweb_df.columns.tolist())
```

Important columns

OBJECT_NAME-        Satellite name
NORAD_CAT_ID-	    Unique satellite catalogue number
EPOCH-	            Time when the orbital data was calculated
INCLINATION	-       Orbital inclination in degrees
ECCENTRICITY-	    Shape of the orbit
MEAN_MOTION	-       Number of orbits per day
RA_OF_ASC_NODE-	    Orientation of the orbital plane
ARG_OF_PERICENTER-	Orientation within the orbit
MEAN_ANOMALY-	    Satellite position within its orbit
BSTAR-	            Atmospheric-drag-related parameter

```python
oneweb_df.to_csv(
    "data/oneweb_satellites.csv",
    index=False
)
```

# startlink constellations

```python
starlink_url = (
    "https://celestrak.org/NORAD/elements/"
    "gp.php?GROUP=STARLINK&FORMAT=CSV"
)

starlink_df = pd.read_csv(starlink_url)
```

# iridium NEXT Constellation

```python
iridium_url = (
    "https://celestrak.org/NORAD/elements/"
    "gp.php?GROUP=IRIDIUM-NEXT&FORMAT=CSV"
)

iridium_df = pd.read_csv(iridium_url)
```

```python
#combine all constellations
oneweb_df["CONSTELLATION"] = "OneWeb"
starlink_df["CONSTELLATION"] = "Starlink"
iridium_df["CONSTELLATION"] = "Iridium NEXT"

satellites_df = pd.concat(
    [
        oneweb_df,
        starlink_df,
        iridium_df
    ],
    ignore_index=True
)

satellites_df.head()
```

```python
satellites_df.to_csv(
    "data/leo_constellations.csv",
    index=False
)
```
